### **Resolución:**

#### 1)

La característica principal de los filtros de Bessel es tener retardo de grupo 
constante, lo que significa que todas las frecuencias en la banda de paso se 
retrasan el mismo tiempo. En este caso realizaremos la aproximación por el 
método de Storch.

Partiendo directamente del comportamiento que queremos para nuestro sistema:

$$
T(s) \approx e^{-sT_0}
$$

Donde $T_0$ es el retardo deseado. Expresando la exponencial en términos de 
funciones hiperbólicas:

$$
T(s) = \frac{1}{\sinh(s) + \cosh(s)}
$$

Utilizando la expansión en fracciones continuas de $\coth(s)$:

$$
\coth(s)
\approx
\frac{1}{s}
+
\frac{1}{\dfrac{3}{s}+\dfrac{1}{\dfrac{5}{s}+\dfrac{1}{\dfrac{7}{s}+\cdots}}}
$$

Podemos determinar las funciones transferencia para cada orden:

$$
\text{Para } N = 1 \qquad T_{B1}(s) = \frac{1}{s + 1}
$$

$$
\text{Para } N = 2 \qquad T_{B2}(s) = \frac{3}{s^2 + 3s + 3}
$$

$$
\text{Para } N = 3 \qquad T_{B3}(s) = \frac{15}{s^3 + 6s^2 + 15s + 15}
$$

$$
\text{Para } N = 4 \qquad T_{B4}(s) = \frac{105}{s^4 + 10s^3 + 45s^2 + 105s + 105}
$$

Notamos que el retardo de grupo en $\omega = 0$ para todos los órdenes es:

$$
D(0) = -\left.\frac{d\angle T(j\omega)}{d\omega}\right|_{\omega=0} = \frac{a_1}{a_0} = 1 \text{ s}
$$

donde $a_0$ y $a_1$ son los coeficientes de menor grado del denominador.

#### 2)
Para hallar el minimo orden recordamos que $\alpha = -20log(|T(s)|) = 10log(|T(s)|^{-2})$ entonces nos queda $\alpha=10log\left(\frac{|B_n(jw)|^2}{|B_n(0)|^2}\right)$,  operando un poco algrebraicamente podemos ir sacando los $\alpha$ tales que cumplan con un $\alpha < \alpha_{max}$ 

$$
\text{Para } N = 2 \qquad \alpha_2 = 10log\left( \frac{2^2+3^2}{3^2}\right) \rightarrow \alpha_2=1,59db > \alpha_{max} \qquad \text{No cumple}.
$$

$$
\text{Para } N = 3 \qquad \alpha_3 = 10log\left( \frac{81+196}{225}\right) \rightarrow \alpha_3=0,9db < \alpha_{max} \qquad \text{Cumple}.
$$

Ya sabemos que para filtros de Bessel con orden mayor o igual a 3, cumple con la plantilla pedida por el enunciado, nos quedamos con el de orden 3.

####  3) 
Evaluamos el retardo de grupo $D(\omega = 2.5)$ y expresamos en forma porcentual 
$(\%)$ el error o desviación respecto a $D(\omega = 0)$.

El retardo de grupo para $N = 3$ es:

$$
D_{B3}(\omega) = \frac{6\omega^4 + 45\omega^2 + 225}{\omega^6 + 6\omega^4 + 45\omega^2 + 225}
$$

Evaluando en $\omega = 2{,}5$:

$$
D_{B3}(2{,}5) = \frac{6(2{,}5)^4 + 45(2{,}5)^2 + 225}{(2{,}5)^6 + 6(2{,}5)^4 + 45(2{,}5)^2 + 225}
$$

$$
\boxed{D_{B3}(2{,}5) = 0{,}752 \text{ s}}
$$

El error porcentual respecto a $D(\omega = 0) = 1\text{ s}$ es:

$$
\text{Error} = \frac{D(2{,}5) - D(0)}{D(0)} \cdot 100
$$

$$
\text{Error} = \frac{0{,}752 - 1}{1} \cdot 100
$$

$$
\boxed{\text{Error} = -24{,}8\%}
$$

En $\omega = 0$ : todas las frecuencias se retrasan 1 segundo. \
En $\omega =2.5$: las frecuencias se retrasan solo 0.752 segundos.

#### 4)
Nos piden un circuito normalizado con estructura Sallen-Key. Sabemos que su transferencia esta caracterizada por la siguiente forma $|T(s)|= \frac{K\cdot \omega_0^2}{s^2+s\cdot\frac{\omega_0}{Q}+\omega_0^2}$ pero como $K=1$ nos deberia quedar $|T(s)|= \frac{\omega_0^2}{s^2+s\cdot\frac{\omega_0}{Q} + \omega_0^2}$. Como sabemos que el filtro de Bessel es de orden 3 y el Sallen-Key es de orden 2 debemos cascadear este sistema con un filtro Pasabajos para poder obtener una transferencia que sea de orden 3.
Entonces sabemos que las transferencias son:
$$
|T_{LP}| = \left( \frac{1}{C_3}\right)\cdot\left(\frac{1}{s+\frac{1}{C_3\cdot R_3}} \right) \qquad \text{Para el RC Pasabajos}
$$
$$
|T_{SK}| = \left( \frac{G_1G_2}{C_1C_2}\right)\cdot\left(\frac{1}{s^2+s\frac{1}{C_2}\cdot\left(G_1+G_2\right)+\frac{G_1G_2}{C_1C_2}} \right) \qquad \text{Para el Sallen-Key}
$$

Entonces ahora necesitamos factorizar nuestro filtro de Bessel para llegar a la forma $|T(s)|= \frac{1}{s+1}\cdot\frac{1}{s^2+s+1}$

Partiendo de:

$$
T_{B3}(s) = \frac{15}{s^3 + 6s^2 + 15s + 15}
$$

Nos termina quedando:

$$
T_{B3}(s) = \frac{1}{s+2,322}\cdot\frac{15}{s^2+s(3,678)+6,46}
$$

Entonces normalizando para que nos cumpla con la transferencia que se mostro anteriormente nos queda:
$$
\text{Para} \qquad G1 = G2 = G3 = 1
$$

$$
    C1 = \frac{1}{C_2\cdot 6,46} \qquad C_2 = \frac{2}{3,678} \qquad C_3=\frac{1}{2,322}
$$

#### **Bonus:**
####  Simulacion en Python


```python
import numpy as np
import matplotlib.pyplot as plt

# ─── Componentes normalizados ───────────────────────────────────────────────
R1, R2, R3 = 1, 1, 1
C1 = 0.2847
C2 = 0.5438
C3 = 0.4307

G1, G2, G3 = 1/R1, 1/R2, 1/R3

# ─── Coeficientes exactos de tu T(s) ────────────────────────────────────────
p1    = 1 / (C3 * R3)          # polo RC
w0_sq = (G1 * G2) / (C1 * C2)  # w0^2 Sallen-Key
coef_s = (G1 + G2) / C2        # coef de s

# Numerador: (1/C3) * (G1G2/C1C2)
NUM_circ = [(1/C3) * w0_sq]

# Denominador: (s + 1/C3R3)(s^2 + coef_s*s + w0_sq)
DEN_circ = [1,
            p1 + coef_s,
            p1 * coef_s + w0_sq,
            p1 * w0_sq]


# ─── Respuesta en frecuencia ────────────────────────────────────────────────
N = 2000
ws = np.logspace(np.log10(0.01), np.log10(10), N)

def freq_resp(w, num, den):
    s = 1j * w
    return np.polyval(num, s) / np.polyval(den, s)

H_circ = np.array([freq_resp(w, NUM_circ, DEN_circ) for w in ws])

mag_circ = 20 * np.log10(np.abs(H_circ))


gd_circ = -np.gradient(np.unwrap(np.angle(H_circ)), ws)

# ─── Valores clave ──────────────────────────────────────────────────────────
D0_c  = np.interp(0.001, ws, gd_circ)
D25_c = np.interp(2.5,   ws, gd_circ)


print(f"\n  Circuito → D(0)={D0_c:.4f}s  D(2.5)={D25_c:.4f}s  Error={abs(D25_c-D0_c)/D0_c*100:.2f}%")

# ─── Graficas ───────────────────────────────────────────────────────────────
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(9, 7), sharex=True)


ax1.semilogx(ws, mag_circ, color='#993C1D', linewidth=1.5, label='Circuito')
ax1.axvline(x=1,   color='green',  linestyle='--', linewidth=1, label='ωp=1')
ax1.axvline(x=2.5, color='orange', linestyle='--', linewidth=1, label='ωs=2.5')
ax1.axhline(y=-1,  color='gray',   linestyle=':',  linewidth=1, label='αmax=1dB')
ax1.set_ylabel('Magnitud [dB]')
ax1.set_title('Bessel N=3 - Circuito')
ax1.legend(fontsize=9)
ax1.grid(True, which='both', alpha=0.3)


ax2.semilogx(ws, gd_circ, color='#993C1D', linewidth=1.5, label=f'Circuito   D(0)={D0_c:.4f}s')
ax2.axvline(x=1,   color='green',  linestyle='--', linewidth=1)
ax2.axvline(x=2.5, color='orange', linestyle='--', linewidth=1)
ax2.scatter([2.5], [D25_c], color='#993C1D', zorder=5, s=60, label=f'D(2.5) Circuito={D25_c:.4f}s')
ax2.set_ylabel('Retardo de grupo [s]')
ax2.set_xlabel('ω [rad/s]')
ax2.legend(fontsize=9)
ax2.grid(True, which='both', alpha=0.3)

plt.tight_layout()
```

#### Desnormalización para $D(0) = 200\ \mu s$

$$s = \frac{s'}{\frac{1}{200\mu s}} \quad \Rightarrow \quad s = \frac{s'}{5000}$$

Mantengo $G_1 = G_2 = G_3 = 1$, entonces:

Considerando la impedancia del capacitor:

$$Z_C = \frac{1}{sC} \quad \Rightarrow \quad Z_C = \frac{1}{s' \cdot \frac{C}{5000}}$$

Desnormalizo

$$C_3' = \frac{C_3}{5000} \qquad C_3 = \frac{1}{2{,}322} = 0{,}43 \qquad \boxed{C_3' = 86\ \mu F}$$

$$C_2' = \frac{C_2}{5000} \qquad C_2 = \frac{2}{3{,}678} = 0{,}5437 \qquad \boxed{C_2' = 108{,}74\ \mu F}$$

$$C_1' = \frac{C_1}{5000} \qquad C_1 = \frac{1}{C_2 \cdot 6{,}46} = 0{,}2846 \qquad \boxed{C_1' = 56{,}92\ \mu F}$$
