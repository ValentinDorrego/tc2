a. Para la red T necesitamos $\dfrac{V_{23}}{V_{13}}$ ; $\dfrac{V_{21}}{V_{31}}$ y $\dfrac{V_{12}}{V_{32}}$ utilizando la matriz **MAI** halla en punto 1).

Para armar los cofactores de segundo orden necesitamos eliminar las filas $mn$ y las columnas $ij$ y realizar el determinante de los elementos que quedan y aplicamos la siguiente formula

$$
A^{ij}_{mn}=\dfrac{V_{ij}}{V_{mn}}
$$

$$
\underline{Y}^{mn}_{ij} = -1^{i+j+m+n}\cdot |\Delta^{mn}_{ij}|
$$
Entonces obtenemos

$$\underline{Y}^{13}_{23} = \frac{1}{ (sL)^2}$$ 
$$\underline{Y}^{13}_{13} = \frac{s^2LC+1}{ (sL)^2}$$ 
$$\boxed{A^{23}_{13} = \frac{ \frac{1}{(sL)^{2}} }{ \frac{s^2LC+1}{ (sL)^2}}=
\frac{1}{LC}\cdot \frac{1}{s^2 + \frac{1}{LC}}}
$$ 

$$\underline{Y}^{31}_{21} = \frac{1}{LC}$$ 
$$\underline{Y}^{31}_{31} = \frac{s^2LC+1}{ (sL)^2}$$ 
$$\boxed{A^{21}_{31} = \frac{ \frac{1}{CL} }{ \frac{s^2LC+1}{ (sL)^2}}=
\frac{s^2}{s^2 + \frac{1}{CL}}}
$$ 

$$\underline{Y}^{32}_{12} = -\frac{C}{L}$$ 
$$\underline{Y}^{32}_{32} = \frac{s^2LC+1}{ (sL)^2}$$ 
$$\boxed{A^{12}_{32} = \frac{ -\frac{C}{L}}{ \frac{s^2LC+1}{ (sL)^2}}=
-\frac{s^2}{s^2 + \frac{1}{LC}}}
$$ 

b. Ahora teniendo $A^{23}_{13}; A^{21}_{31} y A^{12}_{32}$ notamos que las matrices $A^{23}_{13}$ y $A^{21}_{31}$ son complementarias entre si.

Partimos de las expresiones obtenidas:

$$
A_{13}^{23} = \frac{\frac{1}{LC}}{s^2+\frac{1}{LC}}
\qquad\qquad
A_{31}^{21} = \frac{s^2}{s^2+\frac{1}{LC}}
$$

Como ambas comparten el mismo denominador, podemos sumarlas directamente:

$$
A_{23}^{13} + A_{21}^{31}
= \frac{\dfrac{1}{LC}}{s^2+\dfrac{1}{LC}} + \frac{s^2}{s^2+\dfrac{1}{LC}}
= \frac{\dfrac{1}{LC} + s^2}{s^2+\dfrac{1}{LC}}
$$

El numerador es idéntico al denominador, por lo tanto:

$$\boxed{
A_{23}^{13} + A_{21}^{31} = 1 \qquad}$$

**Conclusión:** $A_{23}^{13}$ y $A_{21}^{31}$ son funciones complementarias (pasa-bajos y pasa-altos de un mismo filtro de segundo orden), ya que su suma es la unidad para todo valor de $s$.


**Nota:** en cambio, $A_{12}^{32} = -A_{21}^{3z`1}$, por lo que:

$$
A_{13}^{23} + A_{32}^{12}
= \frac{\dfrac{1}{LC}}{s^2+\dfrac{1}{LC}} - \frac{s^2}{s^2+\dfrac{1}{LC}}
= \frac{\dfrac{1}{LC} - s^2}{s^2+\dfrac{1}{LC}} \neq 1
$$

Esta combinación no es complementaria (corresponde a una función rechaza-banda, no a la unidad).
