**Filtro Comb**
- Una rama pasa directamente.
- Otra rama se retrasa $N$ muestras mediante $z^{-N}$.
- Se multiplica por $c_1$.
- Se resta de la señal original.

Por lo tanto tenemos 
$$
v[n] = x[n] - C_1 \cdot x[n-N]
$$

Aplicando *Transformada Z*

$$V(z)=X(z) - C_1\cdot Z^{-n} \cdot X(z)$$
$$V(z)=X(z) \left(1 - C_1\cdot Z^{-n}\right)$$
$$\boxed{H_{Comb} = \frac{V(z)}{X(z)} = 1 - C_1\cdot Z^{-n}}$$

**Biquad**

Para la parte recursiva de segundo orden, en donde $w[n]$ es la señal interna del sistema. Hay dos retardos $w[n-1] y w[n-2]$, estas se multiplican por $a_1$ y $a_2$ y luego se realimentan
$$a_0 \cdot w[n] = v[n] + a_1\cdot w[n-1] + a_2 \cdot w[n-2]$$
$$a_0 \cdot W(z) = V(z) + a_1\cdot Z^{-1} \cdot W(z) + a_2 \cdot Z^{-2} \cdot W(z)$$

Agrupamos $W(z)$:

$$a_0 W(z) - a_1 z^{-1} W(z) - a_2 z^{-2} W(z) = V(z)$$

$$W(z) \left( a_0 - a_1 z^{-1} - a_2 z^{-2} \right) = V(z)$$

Entonces:

$$\boxed{\frac{W(z)}{V(z)} = \frac{1}{a_0 - a_1 z^{-1} - a_2 z^{-2}}}$$

Parte de salida

La salida se forma con:
* $b_0 w[n]$
* $b_1 w[n-1]$
* $b_2 w[n-2]$

Por lo tanto:

$$y[n] = b_0 w[n] + b_1 w[n-1] + b_2 w[n-2]$$

Transformada Z:

$$Y(z) = b_0 W(z) + b_1 z^{-1} W(z) + b_2 z^{-2} W(z)$$

Factorizamos:

$$Y(z) = W(z) \left( b_0 + b_1 z^{-1} + b_2 z^{-2} \right)$$

Por lo tanto:

$$\frac{Y(z)}{W(z)} = b_0 + b_1 z^{-1} + b_2 z^{-2}$$

*Transferencia del Biquad*

Multiplicamos:

$$\frac{Y(z)}{V(z)} = \frac{Y(z)}{W(z)} \frac{W(z)}{V(z)}$$

Entonces:

$$\boxed{H_{\text{biquad}}(z) = \frac{b_0 + b_1 z^{-1} + b_2 z^{-2}}{a_0 - a_1 z^{-1} - a_2 z^{-2}}}$$

Por lo tanto la transferencia nos termina quedando

$$H(z) = \frac{Y(z)}{V(z)} \cdot \frac{V(z)}{X(z)}$$
$$\boxed{H(z) = \left( 1 - c_1\cdot Z^{-N} \right) \cdot \frac{b_0 + b_1 z^{-1} + b_2 z^{-2}}{a_0 - a_1 z^{-1} - a_2 z^{-2}}}$$

- **¿Es un filtro FIR o un IIR?**


Tenemos la siguiente función de transferencia:

$$H(z) = \frac{1}{3} \frac{1 - z^{-3}}{1 - z^{-1}}$$

A primera vista podría parecer un filtro IIR porque tiene un denominador, pero este es un caso especial donde ocurre una **cancelación de polos y ceros**.

Factorización del numerador
El término del numerador $(1 - z^{-3})$ es una diferencia de cubos que se puede factorizar de la siguiente manera:

$$1 - z^{-3} = (1 - z^{-1})(1 + z^{-1} + z^{-2})$$

Sustituyendo esta factorización en la ecuación original obtenemos:

$$H(z) = \frac{1}{3} \frac{(1 - z^{-1})(1 + z^{-1} + z^{-2})}{1 - z^{-1}}$$

Como el término $(1 - z^{-1})$ se encuentra tanto en el numerador como en el denominador, se simplifican mutuamente (el cero en $z=1$ cancela al polo en $z=1$):

$$H(z) = \frac{1}{3}(1 + z^{-1} + z^{-2})$$

Si distribuimos la constante, la función de transferencia final es:

$$H(z) = \frac{1}{3} + \frac{1}{3}z^{-1} + \frac{1}{3}z^{-2}$$

$$h[n] = \frac{1}{3} x[n] + \frac{1}{3}x[n-1] + \frac{1}{3}x[n-2]$$

Es decir, los coeficientes del filtro son simplemente:
$$h[n] = \left\{\underline{\frac{1}{3}}, \frac{1}{3}, \frac{1}{3}\right\}$$

Entonces el filtro es **FIR**

Esta implementacion es mucho mas eficiente debido a que 

Un FIR de media móvil tradicional:

$$y[n] = \frac{1}{N} (x[n] + x[n-1] + \cdots + x[n-N+1])$$

necesita:
* $N - 1$ retardos.
* Aproximadamente $N$ sumas.
* $N$ multiplicaciones si se implementa directamente.

En cambio, esta estructura usa:

$$\boxed{y[n] = y[n-1] + \frac{1}{N} (x[n] - x[n-N])}$$

Esta ecuación se puede obtener de la estructura.

Entonces solamente necesitás:

* Un retardo de $N$ muestras.
* Una resta.
* Una suma.
* Una multiplicación por $1/N$.


- Usamos:

$$\frac{1 - z^{-N}}{1 - z^{-1}} = 1 + z^{-1} + \cdots + z^{-(N-1)}$$

Si queremos **6 unos**:

$$a_0 = 1, \quad a_1 = 1, \quad c_1 = 1, \quad N = 6, \quad b_0 = 1$$

Entonces:

$$H(z) = \frac{1 - z^{-6}}{1 - z^{-1}}$$

y finalmente:

$$\boxed{H(z) = 1 + z^{-1} + z^{-2} + z^{-3} + z^{-4} + z^{-5}}$$

Por lo tanto:

$$\boxed{h[k] = (1, 1, 1, 1, 1, 1)}$$
