### Resolucion

1.
**Filtro Comb**
- Una rama pasa directamente.
- Otra rama se retrasa $N$ muestras mediante $z^{-N}$.
- Se multiplica por $c_1$.
- Se resta de la señal original.

Por lo tanto tenemos 
$$
v[n] = x[n] - C_1 \cdot x[n-N]
$$

Aplicando *Transformada Z*

$$V(z)=X(z) - C_1\cdot Z^{-n} \cdot X(z)$$
$$V(z)=X(z) \left(1 - C_1\cdot Z^{-n}\right)$$
$$\boxed{H_{Comb} = \frac{V(z)}{X(z)} = 1 - C_1\cdot Z^{-n}}$$

**Biquad**

Para la parte recursiva de segundo orden, en donde $w[n]$ es la señal interna del sistema. Hay dos retardos $w[n-1] y w[n-2]$, estas se multiplican por $a_1$ y $a_2$ y luego se realimentan
$$a_0 \cdot w[n] = v[n] + a_1\cdot w[n-1] + a_2 \cdot w[n-2]$$
$$a_0 \cdot W(z) = V(z) + a_1\cdot Z^{-1} \cdot W(z) + a_2 \cdot Z^{-2} \cdot W(z)$$

Agrupamos $W(z)$:

$$a_0 W(z) - a_1 z^{-1} W(z) - a_2 z^{-2} W(z) = V(z)$$

$$W(z) \left( a_0 - a_1 z^{-1} - a_2 z^{-2} \right) = V(z)$$

Entonces:

$$\boxed{\frac{W(z)}{V(z)} = \frac{1}{a_0 - a_1 z^{-1} - a_2 z^{-2}}}$$

Parte de salida

La salida se forma con:
* $b_0 w[n]$
* $b_1 w[n-1]$
* $b_2 w[n-2]$

Por lo tanto:

$$y[n] = b_0 w[n] + b_1 w[n-1] + b_2 w[n-2]$$

Transformada Z:

$$Y(z) = b_0 W(z) + b_1 z^{-1} W(z) + b_2 z^{-2} W(z)$$

Factorizamos:

$$Y(z) = W(z) \left( b_0 + b_1 z^{-1} + b_2 z^{-2} \right)$$

Por lo tanto:

$$\frac{Y(z)}{W(z)} = b_0 + b_1 z^{-1} + b_2 z^{-2}$$

*Transferencia del Biquad*

Multiplicamos:

$$\frac{Y(z)}{V(z)} = \frac{Y(z)}{W(z)} \frac{W(z)}{V(z)}$$

Entonces:

$$\boxed{H_{\text{biquad}}(z) = \frac{b_0 + b_1 z^{-1} + b_2 z^{-2}}{a_0 - a_1 z^{-1} - a_2 z^{-2}}}$$

Por lo tanto la transferencia nos termina quedando

$$H(z) = \frac{Y(z)}{V(z)} \cdot \frac{V(z)}{X(z)}$$
$$\boxed{H(z) = \left( 1 - c_1\cdot Z^{-N} \right) \cdot \frac{b_0 + b_1 z^{-1} + b_2 z^{-2}}{a_0 - a_1 z^{-1} - a_2 z^{-2}}}$$

- **¿Es un filtro FIR o un IIR?**


Tenemos la siguiente función de transferencia:

$$H(z) = \frac{1}{3} \frac{1 - z^{-3}}{1 - z^{-1}}$$

A primera vista podría parecer un filtro IIR porque tiene un denominador, pero este es un caso especial donde ocurre una **cancelación de polos y ceros**.

Factorización del numerador
El término del numerador $(1 - z^{-3})$ es una diferencia de cubos que se puede factorizar de la siguiente manera:

$$1 - z^{-3} = (1 - z^{-1})(1 + z^{-1} + z^{-2})$$

Sustituyendo esta factorización en la ecuación original obtenemos:

$$H(z) = \frac{1}{3} \frac{(1 - z^{-1})(1 + z^{-1} + z^{-2})}{1 - z^{-1}}$$

Como el término $(1 - z^{-1})$ se encuentra tanto en el numerador como en el denominador, se simplifican mutuamente (el cero en $z=1$ cancela al polo en $z=1$):

$$H(z) = \frac{1}{3}(1 + z^{-1} + z^{-2})$$

Si distribuimos la constante, la función de transferencia final es:

$$H(z) = \frac{1}{3} + \frac{1}{3}z^{-1} + \frac{1}{3}z^{-2}$$

$$h[n] = \frac{1}{3} x[n] + \frac{1}{3}x[n-1] + \frac{1}{3}x[n-2]$$

Es decir, los coeficientes del filtro son simplemente:
$$h[n] = \left\{\underline{\frac{1}{3}}, \frac{1}{3}, \frac{1}{3}\right\}$$

Entonces el filtro es **FIR**

Esta implementacion es mucho mas eficiente debido a que 

Un FIR de media móvil tradicional:

$$y[n] = \frac{1}{N} (x[n] + x[n-1] + \cdots + x[n-N+1])$$

necesita:
* $N - 1$ retardos.
* Aproximadamente $N$ sumas.
* $N$ multiplicaciones si se implementa directamente.

En cambio, esta estructura usa:

$$\boxed{y[n] = y[n-1] + \frac{1}{N} (x[n] - x[n-N])}$$

Esta ecuación se puede obtener de la estructura.

Entonces solamente necesitás:

* Un retardo de $N$ muestras.
* Una resta.
* Una suma.
* Una multiplicación por $1/N$.


- Usamos:

$$\frac{1 - z^{-N}}{1 - z^{-1}} = 1 + z^{-1} + \cdots + z^{-(N-1)}$$

Si queremos **6 unos**:

$$a_0 = 1, \quad a_1 = 1, \quad c_1 = 1, \quad N = 6, \quad b_0 = 1$$

Entonces:

$$H(z) = \frac{1 - z^{-6}}{1 - z^{-1}}$$

y finalmente:

$$\boxed{H(z) = 1 + z^{-1} + z^{-2} + z^{-3} + z^{-4} + z^{-5}}$$

Por lo tanto:

$$\boxed{h[k] = (1, 1, 1, 1, 1, 1)}$$

3. Para poder obtener $y(n)=\frac{x(n)-x(n-2)}{2}$, se debe eliminar *comb* tomando $C_1=0$, luego $a_0=1, b_0= \frac{1}{2}$ y $b_2 = \frac{-1}{2}$
   $$H(z) = \frac{ \frac{1}{2} + \frac{1}{2} z^{-2}}{1} = $$

```python
import numpy as np

fs = 1000
f = 10
dur = 0.1

n = np.arange(int(fs * dur))

Omega = 2 * np.pi * f / fs

# Señal de entrada
x = np.sin(Omega * n)

# Aplicación del filtro
y = np.zeros_like(x)
y[2:] = (x[2:] - x[:-2]) / 2

# Resultado teórico
y_teorica = np.sin(Omega) * np.cos(Omega * (n - 1))

# Error
error = np.max(np.abs(y[2:] - y_teorica[2:]))

print("Error máximo:", error)
```

4. **Dc blocker** para:$a_0=1, a_1=\beta, b_0=1, b_1=-1$ para $\beta = 0.9$

*Coeficientes:*
$$a_0 = 1, \quad a_1 = \beta, \quad b_0 = 1, \quad b_1 = -1 \quad \text{con } \beta = 0.9$$

*Función de transferencia:*
$$H(z) = \frac{b_0 + b_1 z^{-1} + b_2 z^{-2}}{a_0 - a_1 z^{-1} - a_2 z^{-2}} = \frac{1 - z^{-1}}{1 - \beta z^{-1}}$$

$$H(z) = \frac{1 - z^{-1}}{1 - \beta z^{-1}}$$

<div align="center">
  <img src="imagenes/6/Figura3.png" width="600">
  <figcaption>Figura 3: Esquema DC Bloquer.</figcaption>
</div>



Se lo llama *DC Blocker* porque para continua ($\Omega = 0$):

*Evaluando en $z = e^{j\Omega}$:*
$$H(e^{j\Omega}) = \frac{1 - e^{-j\Omega}}{1 - \beta e^{-j\Omega}} = \frac{1 - e^{-j 0}}{1 - \beta e^{-j 0}} = \frac{1 - 1}{1 - \beta} = 0 \quad \longrightarrow \quad \text{Bloquea DC}$$

$\beta$ sirve para controlar qué tan cerca del origen corta el filtro.

- Determinación del parámetro $\beta$ (Atenuación a -3 dB)

Para el filtro $H(z)$ con un cero en $1$ y un polo en $\beta$:
$$H(z) = \frac{1 - z^{-1}}{1 - \beta z^{-1}}$$

Se busca que la respuesta en frecuencia en $\Omega = 0.1\pi$ sea $3\text{ dB}$ menor que en $\Omega = \pi$:

$$20 \log_{10} \left( \frac{|H(e^{j 0.1\pi})|}{|H(e^{j\pi})|} \right) = -3\text{ dB} \quad \implies \quad \frac{|H(e^{j 0.1\pi})|}{|H(e^{j\pi})|} = 10^{-3/20} \approx \frac{1}{\sqrt{2}}$$

Tenemos el módulo al cuadrado $|H(e^{j\Omega})|^2$

$$|H(e^{j\Omega})|^2 = \frac{|1 - e^{-j\Omega}|^2}{|1 - \beta e^{-j\Omega}|^2}$$

* *Numerador:*
  $$|1 - e^{-j\Omega}|^2 = (1 - e^{-j\Omega})(1 - e^{j\Omega}) = 1 - (e^{j\Omega} + e^{-j\Omega}) + 1 = 2 - 2\cos(\Omega)$$

* *Denominador:*
  $$|1 - \beta e^{-j\Omega}|^2 = (1 - \beta e^{-j\Omega})(1 - \beta e^{j\Omega}) = 1 - \beta(e^{j\Omega} + e^{-j\Omega}) + \beta^2 = 1 + \beta^2 - 2\beta\cos(\Omega)$$

 3. Expresión reducida

$$|H(e^{j\Omega})|^2 = \frac{2 - 2\cos(\Omega)}{1 + \beta^2 - 2\beta\cos(\Omega)}$$


Sustituyendo $\cos(\pi) = -1$ en la respuesta en frecuencia:

$$|H(e^{j\pi})|^2 = \frac{2 - 2\cos(\pi)}{1 + \beta^2 - 2\beta\cos(\pi)} = \frac{4}{(1+\beta)^2} \implies |H(e^{j\pi})| = \frac{2}{1+\beta}$$


La condición $20\log_{10}\left(\frac{|H(e^{j0.1\pi})|}{|H(e^{j\pi})|}\right) = -3\text{ dB}$ equivale en potencia a:

$$\frac{|H(e^{j0.1\pi})|^2}{|H(e^{j\pi})|^2} = 10^{-0.3} \approx \frac{1}{2}$$

Igualando los términos:

$$\frac{\frac{2 - 2\cos(0.1\pi)}{1 + \beta^2 - 2\beta\cos(0.1\pi)}}{\frac{4}{(1+\beta)^2}} = \frac{1}{2} \implies (1+\beta)^2 (1 - \cos(0.1\pi)) = 1 + \beta^2 - 2\beta\cos(0.1\pi)$$

Resolvemos la ecuación cuadrática
Sustituyendo $\cos(0.1\pi) \approx 0.951$:

$$(1+\beta)^2 (0.049) = 1 + \beta^2 - 1.902\beta \implies -0.951\beta^2 + 2\beta - 0.951 = 0$$

Raíces obtenidas:
* $\beta_1 = 1.3764$
* $\beta_2 = 0.7265$

Para garantizar la estabilidad del filtro, los polos deben ubicarse dentro del círculo unitario ($|\beta| < 1$):

$$\mathbf{\beta = 0.7265}$$

*Filtro final:*
$$H(z) = \frac{1 - z^{-1}}{1 - 0.7265 z^{-1}}$$

$$H(e^{j\Omega}) = 1 + e^{-j8\Omega} \implies |H(e^{j\Omega})| = |1 + e^{-j8\Omega}|$$
